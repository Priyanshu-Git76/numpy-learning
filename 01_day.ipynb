## Numpy array and Basic


```python
import numpy as np
```

### creating array from list

```python
arr_1d = np.array([1,2,3,4,5,])
print("1D array :", arr_1d)
arr_2d = np.array([[1,2,3],[8,9,0]])
print("2D array :", arr_2d)
```

### List vs numpy array

```python
import numpy as np
py_list = [4,5,6]
print("python list multiplication ", py_list*2) #element wise multiplication

np_array = np.array([1,2,3])
print("python list multiplication ", np_array*2)#eliment wise multiplication


import time
start = time.time()
py_list = [i*2 for i in range(1000000)]
print("\n List opertaion time: ", time.time()-start)

start = time.time()
np_array = np.arange(1000000)*2
print("\n nnumpy opertaion time: ", time.time()-start)

```

## Creating Array from scratch

```python
zeros = np.zeros((3,4))
print("zeors array : \n", zeros)

ones = np.ones((2,3))
print("ones :",ones)

full = np.full((2,2),7)
print("full aray: ", full)

random = np.random.random((2,4))
print("random array :\n", random)

sequence = np.arange(0,12,3)
print("sequence  array :\n", sequence)

```

## Vector , Matrix and Tensor

```python
vector = np.array([1,2,3])
print("Vector:",vector)

matrix = np.array([[1,2,3],
                   [4,5,6]])
print("Matrix:",matrix)

tensor = np.array([[[4,5,6],[7,4,5]],
                   [[4,4,7],[8,2,3]]])
print("Tensor:\n",tensor)
```

### Array properties

```python
arr = np.array([[1,2,3],
                [4,5,6]])
print("shape", arr.shape)
print("Dimension", arr.ndim)
print("Size", arr.size)
print("Data type", arr.dtype)
```

### Array reshaping

```python
arr = np.arange(12)
print("orignal array ",arr)

reshaped = arr.reshape(3,4)
print("\n array reshape ", reshaped)

# Return orignal array copy
flattend = reshaped.flatten()
print("flattend array",flattend)

# Revel (ruturn view, instead of copy)
raveled = reshaped.ravel()
print("raveled array", raveled)

# transpose
transpose = reshaped.T
print("transpose array ", transpose)
```
